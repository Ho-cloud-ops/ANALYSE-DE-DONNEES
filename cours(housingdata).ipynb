```python
## Detecter les valeurs aberantes avec la methode du z-score pour la colonne AGE de HousingData
lim_sup_z_score_age = HousingData['AGE'].mean() + 3*HousingData['AGE'].std()
lim_inf_z_score_age = HousingData['AGE'].mean() - 3*HousingData['AGE'].std()
valeur_aberantes_age = HousingData.loc[(HousingData['AGE'] > lim_sup_z_score_age) | (HousingData['AGE'] < lim_inf_z_score_age)]
display(valeur_aberantes_age)
```

```python
## Detecter les valeurs aberantes avec la methode du z-score pour la colonne CRIM
lim_sup_z_score_crim = HousingData['CRIM'].mean() + 3*HousingData['CRIM'].std()
lim_inf_z_score_crim = HousingData['CRIM'].mean() - 3*HousingData['CRIM'].std()
valeur_aberantes_crim = HousingData.loc[(HousingData['CRIM'] > lim_sup_z_score_crim) | (HousingData['CRIM'] < lim_inf_z_score_crim)]
display(valeur_aberantes_crim)
```

```python
## Remplacement des valeurs aberrantes ou manquantes par la moyenne pour AGE et CRIM
HousingData['AGE'] = HousingData['AGE'].fillna(HousingData['AGE'].mean())
HousingData['CRIM'] = HousingData['CRIM'].fillna(HousingData['CRIM'].mean())
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Sélectionner les colonnes numériques
colonnes_numeriques = HousingData.select_dtypes(include='number').columns

for colonne in colonnes_numeriques:
    # Supprimer les valeurs manquantes
    donnees = HousingData[colonne].dropna()

    # Diviser les données en 4 intervalles
    intervalles = pd.cut(donnees, bins=4)

    # Calculer les effectifs
    repartition = intervalles.value_counts().sort_index()

    # Calculer les pourcentages
    pourcentages = repartition / repartition.sum() * 100

    # Préparer la légende
    legendes = [
        f"{intervalle}: {effectif} observations ({pourcentage:.1f}%)"
        for intervalle, effectif, pourcentage
        in zip(repartition.index, repartition.values, pourcentages.values)
    ]

    # Créer le graphique
    fig, ax = plt.subplots(figsize=(12, 7))

    wedges, texts, autotexts = ax.pie(
        repartition.values,
        labels=None,
        autopct='%1.1f%%',
        startangle=90,
        pctdistance=0.75,
        textprops={'fontsize': 12},
        wedgeprops={'width': 0.45, 'edgecolor': 'white'}
    )

    # Agrandir les pourcentages
    for texte in autotexts:
        texte.set_fontsize(12)
        texte.set_fontweight('bold')

    # Ajouter la légende
    ax.legend(
        wedges,
        legendes,
        title="Répartition des données",
        loc="center left",
        bbox_to_anchor=(1, 0.5),
        fontsize=10
    )

    ax.set_title(
        f"Répartition de la variable : {colonne}",
        fontsize=15,
        fontweight='bold'
    )

    plt.tight_layout()
    plt.show()
```

```python
# Visualisez la répartition de la variable RAD (Accessibilité aux autoroutes)
rad_counts = HousingData['RAD'].value_counts()

total_obs = len(HousingData)
rad_pourcentages = rad_counts / total_obs * 100

plt.figure(figsize=(10, 6))
bars = plt.bar(rad_pourcentages.index.astype(str), rad_pourcentages.values)
plt.xlabel('Indice RAD')
plt.ylabel("Pourcentage d'observations (%)")
plt.title('Répartition de la variable RAD')
plt.xticks(rotation=45)
plt.tight_layout()

for bar in bars:
  yval = bar.get_height()
  plt.text(bar.get_x() + bar.get_width()/2, yval + 0.5, f'{yval:.1f}%', ha='center', va='bottom')

plt.show()
```
